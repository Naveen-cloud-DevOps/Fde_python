```python
def my_function():
    statement
```

```python
def ganesh(): # function name 
    print(" Hello to my world") # function statement 
ganesh()  ## function call 
```

```python
def greet():
    print("Hello Naveen")
    print("Welcome to Python Programming")

greet()

# Here:
#  def – Keyword used to define a function.
#  greet – Function name.
#  () – Parentheses.
#  : – Starts the function body.
#  greet() – Calls the function.
```

```python
# To print statement same using function
def welcome():
    print("Welcome to Python")

welcome()
welcome()
welcome()
```

```python
# To store response from function
def greet():
    print("hello rocky")   
response = greet()
print(response)

# none default result 
```

```python
# You can group multiple deployment operations inside one function.

def deploy():
    print("Starting deployment...")
    print("Building Docker image...")
    print("Pushing image to ECR...")
    print("Deploying application to EKS...")
    
response = deploy()
print(response)
```

```python
def addition():
    a = 5
    b = 40
    total = a+b
    print("first number:", a)
    print("second number:", b)
    print("addition:", total)

addition()
```

```python
def greet():
    print("hello ganesh")
for i in range (1, 4):
    print("Iteration:", i)
    greet()
```

## 1.2 No Parameters with return value

```python
def shiva():
    annaul_income = 20000
    monthly_income =  annaul_income / 12
    return monthly_income
result =shiva()
print(result)
```

```python
def get_age():
    age = 25
    return (age)
user_age = get_age()
if user_age >= 18:
    print("major")
else:
    print("minor")
```

```python
def caliculate_final():
    price = 1000
    discount = 10
    final_price = price -(price * discount / 100)
    return final_price
caliculate_final()
```

```python
# without function
### first person salary ###
salary1 = 70000
monthly1 = salary1 / 12 
print(monthly1)

### second person salary ###
salary2 = 140000
monthly2 = salary2 / 12 
print(monthly2)

### third person salary ###
salary3 = 270000
monthly3 = salary3 / 12 
print(monthly3)
```

```python
# with function 
def calculate_monthly_salary(salary):
    monthly = salary / 12
    return monthly 
result1=calculate_monthly_salary(60000)
print(result1)
result2=calculate_monthly_salary(120000)
print(result2)
result3=calculate_monthly_salary(240000)
print(result3)
```
