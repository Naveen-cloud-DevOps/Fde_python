# oops : Object Orientd Prograaming
### oops is programming approach where we organise our code using classes and objects
### Class is blue print or template to create objects

### Class  blue print of the object
### Objects instance of the class

```python
# basic syntax of class and object in python
class ClassName: # class is python keyword and ClassName is the name of the class
   variable1 = "value1" #variable values
   variable2 = "value2"
   variable3 = "value3"
c2 = ClassName() # object of the class

```

```python
class Student:
    name = "Ganesh"
    rollno = 123
    age = 20
s1 = Student() # object of the class
print(Student.name) # accessing class variable using class name # without reference to s1 object
print(Student.rollno) # accessing class variable using class name
print(Student.age) # accessing class variable using class name
print(s1.name) # accessing class variable using object name # with reference to s1 object
print(s1.rollno) # accessing class variable using object name   
print(s1.age) # accessing class variable using object name  
```

## Class variables
### A class variable is a variable declared directly inside a class but outside its methods. It is shared by all objects of that class by default.

```python
#Class variables as defined inside the class and can be accessed using object of the class.
class Student:  
    name = "veera" 
    roll = 232
    age = 20
```

## Instance Variables
### An instance variable is a variable that belongs to an individual object (instance) of a class.

```python
# example class variables and instance variable 
class Student:
    school = "ABC" #class variable as deffines inside the class
    country = "India"
    #insatance method
    def display(self, name, roll, age): #self is a reference to the current instance of the class and is used to access variables that belongs to the class.
      self.name = name #instance variable as deffines inside the fucntion method
      self.roll = roll
      self.age = age
s1=Student() # creation of object of class Student
s1.display("veera", 232, 20)
#class variable can be accessed using object of the class
print(s1.school) #accessing class variable using object of class Student
print(s1.country) #accessing class variable using object of class Student

##instance varibles can be accessed using object of the class
print(s1.name) #accessing instance variable using object of class Student
print(s1.roll) #accessing instance variable using object of class Student
print(s1.age) #accessing instance variable using object of class Student
```
