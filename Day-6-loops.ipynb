```python
# for loop with dictionary
#for loop calling total items 
number = {
    "name" : "ganesh",
    "course" : "Fde"
}
for i in number.items():
    print (i)
    
```

```python
## for loop with dictionry 
## for loop calling values 

number = {
    "name" : "ganesh" ,
    "course" : "Fde"
}
for i in number.values():
    print (i)
```

```python
# for loop calling keys 
number = {
    "name"  : "ganesh",
    "course": "Fde"
}
for i in number.keys():
    print(i)
```

```python
import random

for i in range(1, 2):
    num = random.randint(1, 200)

    if num % 2 == 1:
        print(f"Number found {num}, Iteration: {i}")
        break

    print(f"Try again: {num}, Iteration: {i}")

else:
    print("No odd number found in 4 attempts")
#random.randint() is a built-in function of Python's random module
```

```python
import random

for i in range(1, 6):

    num = random.randint(1, 100)

    if num % 2 == 0:
        print("Even number:", num, "| Iteration:", i)

    if num % 2 != 0:
        print("Odd number:", num, "| Iteration:", i)

    if num > 50:
        print("Number is greater than 50:", num)
```
